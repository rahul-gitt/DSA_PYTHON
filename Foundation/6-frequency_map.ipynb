```python
n = list(map(int, input("Enter Elements : ").split()))

freq_map = {}

for i in range(0, len(n)):
    if n[i] in freq_map:
        freq_map[n[i]]+= 1
    else:
        freq_map[n[i]] = 1

print(f"Number Frequency is {freq_map}")
```

```python
nums = list(map(int, input("Enter elements : ").split()))

n= len(nums)
hash_map = {}

for i in range(0, n):
    hash_map(nums[i]) = hash_map.get(nums[i], 0) +1

print(hash_map)
```
