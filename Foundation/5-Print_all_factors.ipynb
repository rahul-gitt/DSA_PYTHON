Brut Force Solution:

```python
n = int(input("Enter a  number : "))
num = n
result = []

for i in range(1, num+1):
    if num% i == 0:
        result.append(i)

print(f"The Factor's are {result}.")
```

Avarage Solution : 

```python
n = int(input("Enter a number : "))
num = n
result = []

for i in range(1, (num//2)+1):
    if num % i == 0:
        result.append(i)
result.append(num)

print(f"The Factor's are {result}.")
```
