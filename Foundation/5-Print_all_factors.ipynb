Brut Force Solution:

```python
n = int(input("Enter a  number : "))
num = n
result = []

for i in range(1, num+1):
    if num% i == 0:
        result.append(i)

print(f"The Factor's are {result}.")
```

Avarage Solution : 

```python
n = int(input("Enter a number : "))
num = n
result = []

for i in range(1, (num//2)+1):
    if num % i == 0:
        result.append(i)
result.append(num)

print(f"The Factor's are {result}.")
```

TC = O(n/2)  = O(n)
SC = O(d)     d = number of digits

Optimal Solution

```python
from math import sqrt

n = int(input("Enter a number : "))
num = n
result = []

for i in range(1,int(sqrt(num)+1)):
    if num % i == 0:
        result.append(i)
        if num // i != i:
            result.append(num//i)

result.sort()
print(f"The  numbers are = {result}")
```

TC = O(√n log n)
SC = O(k)  , k = ammount of Factors
